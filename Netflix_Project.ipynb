```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('/content/combinedNetflixData (1).txt',header=None,names=['CustID','Ratings'],usecols=[0,1])
```

```python
df
```

```python
df.isnull().sum()
```

```python
movie_count = df.isnull().sum()
movie_count = movie_count['Ratings']
movie_count
```

```python
#To calculate how many unique user there
total_count=df['CustID'].nunique()
total_count
```

```python
customer_count=total_count-movie_count
customer_count
```

```python
# how much rating count there
rating_count=df['CustID'].count()-movie_count
rating_count
```

```python
#Let's Just make a clear DataFrame

movie_id=None
movie_np=[]

#Iterate through CustID column
for i in df['CustID']:
  if ":" in i:
    #Udate the current movieid in column
    movie_id=int(i.replace(':',''))
  movie_np.append(movie_id)
```

```python
movie_np
```

```python
df['MovieID']=movie_np
```

```python
df
```

```python
df=df[df['Ratings'].notna()]
```

```python
df
```

```python
df.info()
```

```python
df['CustID']=df['CustID'].astype(int)
df
```

```python
df.info()
```

```python
#Pre-Filtering
#Remove all those users who has watched less movie
#Remove all those movies who has less ratings given by the people
```

```python
df_movie_summary=df.groupby('MovieID')['Ratings'].agg(['count'])
df_movie_summary
```

```python
#Now we will create a Benchmark(consider a becnhmark value of 60 percentile)
movie_benchmark=round(df_movie_summary['count'].quantile(0.6),0)
movie_benchmark
```

```python
#If any movie got number of ratings less than 908 people i will remove that movie
```

```python
drop_movie_list=df_movie_summary[df_movie_summary['count']<movie_benchmark].index
drop_movie_list
```

```python
df_customer_summary=df.groupby('CustID')['Ratings'].agg(['count'])
df_customer_summary
```

```python
customer_benchmark=round(df_customer_summary['count'].quantile(0.6),0)
customer_benchmark
```

```python
drop_customer_list=df_customer_summary[df_customer_summary['count']<customer_benchmark].index
drop_customer_list
```

```python
len(drop_customer_list)
```

```python
len(drop_movie_list)
```

```python
df=df[~df['CustID'].isin(drop_customer_list)]
df=df[~df['MovieID'].isin(drop_movie_list)]
```

```python
df
```
